```python
def insertion_sort(array):
    for index in range(1, len(array)):
        key = array[index]
        previous = index - 1
        while previous >= 0 and array[previous] > key:
            array[previous + 1] = array[previous]
            previous = previous - 1
        array[previous + 1] = key
    return array
```

```python
Array = [5,4,3,2,1]
print(insertion_sort(Array))
```
